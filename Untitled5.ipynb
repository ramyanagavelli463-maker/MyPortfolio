```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("retail_sales_dataset.csv.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
missing = df.isnull().sum()

print("Missing values in each column:")
print(missing)
```

```python
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
df = df.drop_duplicates()

print("Duplicates removed successfully!")
print("New dataset shape:", df.shape)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include='object').columns

# Fill numerical missing values with median
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode
for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values handled successfully!")
print(df.isnull().sum())
```

```python
df['transaction_date'] = pd.to_datetime(
    df['transaction_date'],
    errors='coerce'
)

print(df['transaction_date'].dtype)
```

```python
print("Final shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

df.head()
```

```python
df['Year'] = df['transaction_date'].dt.year
df['Month'] = df['transaction_date'].dt.month
df['Month_Name'] = df['transaction_date'].dt.month_name()

print(df[['transaction_date', 'Year', 'Month', 'Month_Name']].head())
```

```python
print(df.columns.tolist())
```

```python
total_sales = df['sales_amount'].sum()

print("Total Sales:", total_sales)
```

```python
monthly_sales = df.groupby(
    ['Year', 'Month', 'Month_Name']
)['sales_amount'].sum().reset_index()

monthly_sales
```

```python
# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

print(df.columns.tolist())
```

```python
# Show the actual columns first
print("Available columns:")
print(df.columns.tolist())

# Find the product category column automatically
category_col = None
for col in df.columns:
    name = col.lower().replace(" ", "_")
    if "product" in name and "categor" in name:
        category_col = col
        break

# Find the sales/amount column automatically
amount_col = None
for col in df.columns:
    name = col.lower().replace(" ", "_")
    if "total" in name and "amount" in name:
        amount_col = col
        break

print("\nCategory column found:", category_col)
print("Amount column found:", amount_col)

# Create category sales report
if category_col is not None and amount_col is not None:
    category_sales = df.groupby(category_col)[amount_col].sum().reset_index()

    category_sales = category_sales.sort_values(
        by=amount_col,
        ascending=False
    )

    print("\nCategory Sales Report:")
    print(category_sales)
else:
    print("\nColumn not found. Please check the column names printed above.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Duplicates removed:", before - after)
print("Rows after removing duplicates:", len(df))
```

```python
# Fill numerical columns with median
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

# Fill text/categorical columns with mode
categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:
    if df[col].isnull().any():
        mode_value = df[col].mode()

        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value.iloc[0])

print("Missing values handled successfully!")
```

```python
print("Total missing values:", df.isnull().sum().sum())
print("Total duplicate rows:", df.duplicated().sum())

print("\nData after cleaning:")
print(df.head())
```

```python
date_col = None

for col in df.columns:
    name = col.lower().replace(" ", "_")

    if "date" in name:
        date_col = col
        break

print("Date column found:", date_col)
```

```python
if date_col is not None:
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')

    print("Date column converted successfully!")
    print(df[date_col].head())
else:
    print("No date column found.")
```

```python
if date_col is not None:

    df['Year'] = df[date_col].dt.year
    df['Month'] = df[date_col].dt.month
    df['Month_Name'] = df[date_col].dt.month_name()

    print(df[[date_col, 'Year', 'Month', 'Month_Name']].head())
```

```python
amount_col = None

possible_names = [
    'total_amount',
    'sales_amount',
    'amount',
    'total_sales',
    'sales',
    'revenue'
]

for col in df.columns:
    name = col.lower().replace(" ", "_")

    if name in possible_names:
        amount_col = col
        break

# If exact name was not found, search more generally
if amount_col is None:
    for col in df.columns:
        name = col.lower().replace(" ", "_")

        if ('amount' in name or 'sales' in name or 'revenue' in name):
            amount_col = col
            break

print("Sales/Amount column found:", amount_col)
```

```python
if amount_col is not None:

    total_sales = df[amount_col].sum()
    average_sales = df[amount_col].mean()
    maximum_sales = df[amount_col].max()
    minimum_sales = df[amount_col].min()

    print("===== SALES SUMMARY =====")
    print("Total Sales:", total_sales)
    print("Average Sales:", average_sales)
    print("Maximum Sales:", maximum_sales)
    print("Minimum Sales:", minimum_sales)

else:
    print("Sales column was not found.")
```

```python
if date_col is not None and amount_col is not None:

    monthly_sales = (
        df.groupby('Month_Name')[amount_col]
        .sum()
        .reset_index()
    )

    print("Monthly Sales Report:")
    print(monthly_sales)

else:
    print("Required columns not found.")
```

```python
category_col = None

for col in df.columns:
    name = col.lower().replace(" ", "_")

    if "product" in name and "categor" in name:
        category_col = col
        break

print("Product category column:", category_col)

if category_col is not None and amount_col is not None:

    category_sales = (
        df.groupby(category_col)[amount_col]
        .sum()
        .reset_index()
        .sort_values(by=amount_col, ascending=False)
    )

    print("\n===== CATEGORY SALES REPORT =====")
    print(category_sales)

else:
    print("Category or sales column not found.")
```

```python
gender_col = None

for col in df.columns:
    name = col.lower().replace(" ", "_")

    if 'gender' in name or 'sex' in name:
        gender_col = col
        break

print("Gender column:", gender_col)

if gender_col is not None and amount_col is not None:

    gender_sales = (
        df.groupby(gender_col)[amount_col]
        .sum()
        .reset_index()
        .sort_values(by=amount_col, ascending=False)
    )

    print("\n===== GENDER SALES REPORT =====")
    print(gender_sales)

else:
    print("Gender or sales column not found.")
```

```python
if category_col is not None and amount_col is not None:

    plt.figure(figsize=(10, 6))

    plt.bar(
        category_sales[category_col],
        category_sales[amount_col]
    )

    plt.title("Sales by Product Category")
    plt.xlabel("Product Category")
    plt.ylabel("Total Sales")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()
```

```python
if date_col is not None and amount_col is not None:

    plt.figure(figsize=(12, 6))

    plt.bar(
        monthly_sales['Month_Name'],
        monthly_sales[amount_col]
    )

    plt.title("Monthly Sales")
    plt.xlabel("Month")
    plt.ylabel("Total Sales")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()
```

```python
if gender_col is not None and amount_col is not None:

    plt.figure(figsize=(8, 5))

    plt.bar(
        gender_sales[gender_col],
        gender_sales[amount_col]
    )

    plt.title("Sales by Gender")
    plt.xlabel("Gender")
    plt.ylabel("Total Sales")

    plt.tight_layout()
    plt.show()

```
